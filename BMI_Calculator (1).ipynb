# BMI Calculator Python beginner project 

```python
name = input("Enter your name:")
weight = float(input("Enter your body weight in kg:"))
height = float(input("Enter your height in cm:"))
bmi = (weight * 10000) / (height * height)
print("Your BMI is:", round(bmi, 2))
if bmi > 0:
    if(bmi < 18.5):
        print(name + ",your Under weight")
    elif (bmi <= 24.9):
        print(name + ",your Normal weight")
    elif (bmi <=29.9):
        print(name + ",your Over weight")
    elif (bmi <=34.9):
        print(name + ",your Obese")
    elif (bmi <=39.9):
        print(name + ",your Severely Obese")
    else:
        print(name + ",your Morbidly Obese") 
else:
    print("Enter a valid input")
    
```

```python
Under 18.5	Underweight	Minimal
18.5 - 24.9	Normal Weight	Minimal
25 - 29.9	Overweight	Increased
30 - 34.9	Obese	High
35 - 39.9	Severely Obese	Very High
40 and over	Morbidly Obese	Extremely High
```
